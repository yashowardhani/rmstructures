```python
from prettytable import PrettyTable
path = '.'
p = np.load('%s/alpha_psamples.npz'%path, allow_pickle=True)
plt.rcParams["font.family"] = "serif"
```

```python
setA_nrmse = p['nrmse'][p['old_nrmse']==None]
setB_nrmse = p['old_nrmse'][p['old_nrmse']!=None]
setB_nrmse_improved = p['nrmse'][p['old_nrmse']!=None]
```

```python
plt.figure(figsize=(6,5))
plt.hist(setA_nrmse, bins=60, label='set A fits', range=(0,0.5))
plt.hist(setB_nrmse, bins=60, label='set B fits', range=(0,0.5))
plt.hist(setB_nrmse_improved, bins=60, label='improved set B fits', range=(0,0.5))
plt.xlabel('normalised rms error')
plt.ylabel('number of datasets')
plt.legend()
plt.savefig('nrmse.png')
plt.savefig('nrmse.pdf')
```
